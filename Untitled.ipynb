type - тип недвижимости

sub_type - подтип недвижимости

start_date - дата начала размещения на рынке

end_date - дата, когда объявление перестает быть активным на рынке

listing_type - тип объявления: 
    - Satılık = Продажа
    - Kiralık = Аренда
    
tom - срок экспозиции на рынке

building_age - возраст недвижимости

total_floor_count - общее кол-во этажей

room_count - кол-во комнат

size - размер дома

addres - адрес

furnished - с мебелью:
    - Eşyalı = С мебелью, 
    - Eşyasız = Без мебели, 
    - Sadece Beyaz Eşya = только белая бытовая техника, 
    - Sadece Mutfak = только кухня
    
currency - валюта цены на жильё

floor_no - Информация о номере этажа данного объявления:
    - Yüksek giriş = 1st Floor
    - Гириш кат, Кот1, Земин кат = первый этаж
    - Çatı Katı = пентхаус
    - Bahçe katı = квартира на этаже в саду

price - цена недвижимости

heating_type - тип обогревательной системы:
    - Kalorifer (Doğalgaz) = central heating/natural gas, 
    - Kalorifer (Kömür) = central heating/coal, 
    - Kombi (Elektrikli) = combi boiler 
    - Klima = air-conditioning 
    - Kombi (Doğalgaz) = gas-combi boiler
    - Merkezi Sistem = central heating 
    - Merkezi Sistem (Isı Payı Ölçer) =central heating
    - Yerden Isıtma = floor heating
    - Soba (Kömür) = heating stove/coal,
    - Soba (Doğalgaz) = heating stove/natural gas
    - Güneş Enerjisi = solar energy,
    - Jeotermal = geothermal energy
    - Fancoil = a type of air conditioner
    - Kat Kaloriferi = central heating
    - Kalorifer (Akaryakıt) = central heating/fuel oil
    - Yok = None
